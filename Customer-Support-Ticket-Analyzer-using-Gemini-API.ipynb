# 🎫 Customer Support Ticket Analyzer using Gemini API

## 📌 Problem Statement

Ek company ko customers se email aur chat ke through support messages milte hain.

Hume Python aur Gemini API ka use karke ek **Customer Support Ticket Analyzer** banana hai jo customer ke message ko analyze karke ye 5 cheezein bataye:

1. Category
2. Sentiment
3. Priority
4. Professional response
5. Next action for support team

Output **structured JSON** format me hona chahiye.

---

# 🔄 Overall Flow

Customer Message
       ↓
Python Program
       ↓
Gemini API
       ↓
System Instruction
       ↓
Gemini Analysis
       ↓
Response Schema
       ↓
Structured JSON
       ↓
Category + Sentiment + Priority + Response + Next Action

````python
# required Libraries Import Karo
from google import genai
from google.genai import types
from google.colab import userdata
import json
````

````python
# API Key Setup
api_key = userdata.get("GEMINI_API_KEY")
````

````python
# Create Gemini Client
client =genai.Client(api_key=api_key)
````

````python
# Customer Message
customer_message = """
I was charged twice for the same order.

I have already emailed support twice and nobody
has responded. I need this fixed immediately.
"""
````

````python
# System Instruction
system_instruction = """
You are a professional customer support ticket analyzer.

Analyze the customer's message and identify:
1. Category
2. Sentiment
3. Priority
4. Professional customer response
5. Next action for the support team

Be accurate, concise, and professional.
"""
````

````python
# Response Schema
response_schema = {
    "type": "object",
    "properties": {
        "category": {
            "type": "string"
        },
        "sentiment": {
            "type": "string"
        },
        "priority": {
            "type": "string"
        },
        "response": {
            "type": "string"
        },
        "next_action": {
            "type": "string"
        }
    },
    "required": [
        "category",
        "sentiment",
        "priority",
        "response",
        "next_action"
    ]
}
````

````python
# Gemini API Call
response = client.models.generate_content(
    model="gemini-3.5-flash",
    contents=customer_message,
    config=types.GenerateContentConfig(
        system_instruction=system_instruction,
        response_mime_type="application/json",
        response_schema=response_schema,
        thinking_config=types.ThinkingConfig(
            thinking_level="low"
        ),
        max_output_tokens=300
    )
)
````

````python
# Response Print
print(response.text)
````

````python
# Reusable Code
def analyze_ticket(customer_message):

    response = client.models.generate_content(
        model="gemini-3.5-flash",
        contents=customer_message,
        config=types.GenerateContentConfig(
            system_instruction=system_instruction,
            response_mime_type="application/json",
            response_schema=response_schema,
            thinking_config=types.ThinkingConfig(
                thinking_level="low"
            ),
            max_output_tokens=300
        )
    )

    return response.text
````

````python
result = analyze_ticket(customer_message)

print(result)
````

````python
# Convert Json to Python Dictionary
result = analyze_ticket(customer_message)

data = json.loads(result)

print(data["category"])
print(data["sentiment"])
print(data["priority"])
print(data["response"])
print(data["next_action"])
````

````python
messages = [

"""
I was charged twice for my subscription this month.

I have checked my bank statement and both transactions
were completed. Please refund the duplicate charge.
""",

"""
The application crashes every time I try to upload a PDF.

I have restarted my computer and reinstalled the application,
but the problem is still happening.
""",

"""
I forgot my password and requested a password reset three
times, but I still haven't received the reset email.

I need access to my account urgently.
""",

"""
My order was supposed to arrive three days ago, but the
tracking information hasn't changed since it left the
warehouse.

Can you please tell me where my package is?
""",

"""
I would like to know whether you offer a student discount
and what documents are required to qualify for it.
"""
]
````

````python
# Multiple Tickets Analyze
for i, message in enumerate(messages, start=1):

    print(f"\nTicket {i}")
    print("-" * 50)

    result = analyze_ticket(message)

    print(result)
````

````python
# Structured Output
for i, message in enumerate(messages, start=1):

    result = analyze_ticket(message)

    data = json.loads(result)

    print(f"\nTicket {i}")
    print("-" * 50)

    print("Category:", data["category"])
    print("Sentiment:", data["sentiment"])
    print("Priority:", data["priority"])
    print("Response:", data["response"])
    print("Next Action:", data["next_action"])
````
